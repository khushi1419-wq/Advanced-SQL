**Q1. What is a Common Table Expression (CTE), and how does it improve SQL query readability?**

Answer:

A CTE is a temporary named result set created using the WITH clause. It exists only during the execution of a single SQL query.

It improves readability:

1. Breaks complex queries into smaller, logical parts.
2. Makes SQL code easier to read and understand.
3. Reduces repeated subqueries.
4. Simplifies debugging and maintenance.
5. Supports recursive queries for hierarchical data.


**Q2. Why are some views updatable while others are read-only? Explain with an example.**

Answer:

A view is updatable if it is based on a single table and does not contain complex operations such as JOIN, GROUP BY, aggregate functions (SUM, COUNT, etc.), DISTINCT, or UNION. In such cases, changes made to the view are reflected in the original table. A view becomes read-only when it includes these complex operations because the database cannot determine how to update the underlying data correctly.

**Example:** A view created as SELECT EmployeeID, Name FROM Employees; is updatable because it is based on a single table. However, a view created as SELECT Department, COUNT(*) FROM Employees GROUP BY Department; is read-only because it uses the GROUP BY clause and an aggregate function.

**Q3. What advantages do stored procedures offer compared to writing raw SQL queries repeatedly?**

Answer:

Stored procedures are pre-written SQL programs stored in the database that can be executed whenever needed. They offer several advantages over writing raw SQL queries repeatedly. They reduce code duplication by allowing the same SQL code to be reused multiple times. Stored procedures improve performance because they are precompiled and executed more efficiently. They also provide better security by restricting direct access to database tables. Additionally, they make database applications easier to maintain, as any changes to the SQL logic need to be made only once in the stored procedure.

**Q4. What is the purpose of triggers in a database? Mention one use case where a trigger is essential.**

Answer:

A trigger is a special type of stored program that automatically executes when a specific event, such as INSERT, UPDATE, or DELETE, occurs on a table. Triggers are used to enforce business rules, maintain data integrity, and automate database tasks without requiring manual intervention.

Example: A trigger can automatically record changes made to an employee's salary in an audit table whenever the salary is updated. This helps maintain a history of changes for security and tracking purposes.

**Q5. Explain the need for data modelling and normalization when designing a database.**

Answer:

Data modelling is the process of designing the structure of a database, including tables, relationships, and constraints. It is needed to organize data properly, ensure consistency, and understand how different data elements are related before building the actual database. It helps in creating an efficient and clear database design.

Normalization is the process of organizing data in a database to reduce redundancy and improve data integrity. It involves dividing large tables into smaller related tables and linking them using relationships. This helps avoid data duplication, prevents update anomalies, and ensures data is stored in a consistent way.

Together, data modelling and normalization help in creating a well-structured, efficient, and reliable database system.
